## What is Matplotlib
* It is a python library for creating static and interective visualization in python.
* Essential for data visualization, data analysis and scientific computing.
* works seamlessly with numpy and pandas.

First import the lib.
If you have not installed this lib, type `pip install matplotlib`

```python
import matplotlib.pyplot as plt
import numpy as np 
import pandas as pd
```

```python
#for jupyter notebook use this magic command to automatically disply plots inline 
%matplotlib inline 
```

## Matplotlib architecture
* **figure** The top level container (the entire window/page)
* **Axes** The actual plot area (where data is drawn)
* **Axis** The number of line Objects(x-axis, y-axis)

Let's start out fisrt plotting. 
Which in line graph

```python
# Method 1: pyplot approach
x = [1,2,3,4,5]
y = [2,4,6,8,10]

plt.plot(x,y) #this will plot the line graph
plt.xlabel('numbers (n)') # this will give the axis a lable(name)
plt.ylabel('n X 2')
plt.title("simple line graph") # this will give the graph a name
plt.show() # this will plot the graph, but in jupyternotebook it display automaticaly but includiing it is good practice.
```

```python
# Method2: Object oriented approach (more control good for complex plotting.)
fig, ax = plt.subplots() # creates and figure and axes objects.
ax.plot(x,y)
ax.set_xlabel('numbers (n)')
ax.set_ylabel('n X 2')
ax.set_title('Line plot (Object-Oriented)')
plt.show()
```

## Basic line plot

```python
# create some sample data
x = np.linspace(0,10,100)
y = np.sin(x)
plt.figure(figsize=(8,5)) # width and height
plt.plot(x,y)
plt.xlabel('X values')
plt.ylabel('Sine values for X')
plt.title('Sine graph')
plt.grid(True)
plt.show()
```

```python
# create some sample data
x = np.linspace(0,10,100)
y1 = np.sin(x)
y2 = np.cos(x)
plt.figure(figsize=(8,5)) # width and height
plt.plot(x,y1,label='sin(x)')
plt.plot(x,y2,label='cos(x)')
plt.xlabel('X values')
plt.ylabel('Y values')
plt.title('sin and cos graph')
plt.legend() # show legend(blue line and orange line specified you are seeing in the image)
plt.grid(True)
plt.show()
```

Customization of a graph

```python
x= np.linspace(0, 10, 20)

# Different line styles and markers
plt.figure(figsize=(12,6))

plt.plot(x,x,'b-',label='Solid line', linewidth=2)
plt.plot(x, x+1, 'r--', label='Dashed line', linewidth=2)
plt.plot(x,x+2, 'g-.', label='Dash-dot line', linewidth=2)
plt.plot(x, x+3, 'm:', label='Dotted line', linewidth=2)
plt.plot(x, x+4, 'co-', label='Circles', markersize=8)
plt.plot(x, x+5, 's-', label='Squares', markersize=6)
plt.plot(x, x+6, '^-', label='Triangles', markersize=6)

plt.xlabel('X Axis')
plt.ylabel('Y Axis')
plt.title('Line Styles and Markers')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

## Scatter plot

```python
# generate random data
np.random.seed(42)
x = np.random.randn(100)
y = 2 * x + np.random.randn(100) * 0.5 # linear relationship with noise

# basic scatter plot

plt.figure(figsize=(8,6))
plt.scatter(x,y, alpha=0.6) # alpha controls transparency
plt.xlabel('x_values')
plt.ylabel('y_values')
plt.title('scatter plot')
plt.grid(True)
plt.show()
```

```python
# Scatter plot with different colors
np.random.seed(42)
x = np.random.randn(100)
y = np.random.randn(100) 
colors = np.random.randn(100)
sizes = 1000 * np.random.randn(100)

plt.figure(figsize=(10,6))
plt.scatter(x,y, c=colors, s=sizes, alpha=0.6, cmap='viridis')
plt.colorbar(label='Color intensity') # add colorbar
plt.xlabel('X_values')
plt.ylabel('Y_values')
plt.title("Colored scatter plots")
plt.grid(True)
plt.show()
```

## Bar chart

```python
categories = ['A','B','C','D','E']
values = [1,2,3,4,5]

# plotting
plt.figure(figsize=(8,6))
plt.bar(categories, values, linewidth=1.2, edgecolor='black')
plt.xlabel('categories')
plt.ylabel('values')
plt.title("Bar chart")
plt.grid(True, alpha=0.2, color='blue')
plt.show()
```

```python
# horizontel bar chart

plt.figure(figsize=(8,6))
plt.barh(categories, values, color='coral', edgecolor='black', linewidth=1.2)
plt.xlabel('Values')
plt.ylabel('Categories')
plt.title('Horizontal Bar Chart')
plt.grid(True, alpha=0.3, axis='x')
plt.show()
```

```python
# Grouped bar chart
categories = ['Q1', 'Q2', 'Q3', 'Q4']
product_a = [20, 35, 30, 35]
product_b = [25, 32, 34, 20]
product_c = [30, 25, 30, 25]

x = np.arange(len(categories))
width = 0.25 # Width of bars

plt.figure(figsize=(10, 6))
# plt.bar(x, height, width, ... ) parameters:
# x - width: x-position (shift left by one bar width to position first group)
# product_a: height values for Product A bars
# width: width of each bar (0.25)
# label: text for legend
# color: bar fill color (hex code)
plt.bar(x - width, product_a, width, label='Product A', color='#FF6B6B')
# x: center position for second group (no shift)
plt.bar(x, product_b, width, label='Product B', color='#4ECDC4')
# x+ width: x-position (shift right by one bar width to position third group)
plt.bar(x + width, product_c, width, label='Product C', color='#95E1D3')

plt.xlabel('Quarter')
plt.ylabel('Sales')
plt.title('Quarterly Sales Comparison')
plt.xticks(x, categories)
plt.legend()
plt.grid(True, alpha=0.3, axis='y')
plt.show()
```

## Histogram

```python
# generate sample data
np.random.seed(42)
data = np.random.normal(100,15,1000) # mean = 100, std=15, datapoints = 1000

# Basic histogram
plt.figure(figsize=(10,6))
plt.hist(data, bins=30, color='skyblue', edgecolor='black', alpha=0.7)
plt.xlabel('Value')
plt.ylabel('Frequency')
plt.title('Histogram of normal data')
plt.grid(True)
plt.show()
```

```python
# multiple histogram
np.random.seed(42)
data1 = np.random.normal(100,15,1000)
data2 = np.random.normal(120,20,1000)

#plotting
plt.figure(figsize=(10,6))
plt.hist(data1, bins=30, label='Data_set 1', color='blue', alpha=0.7)
plt.hist(data2, bins=30, label='Data_set 2', color='red', alpha=0.7)
plt.xlabel("value")
plt.ylabel("frequency")
plt.title('Overlapping Histogram')
plt.legend()
plt.grid(True, axis="y", alpha=0.3)
plt.show()
```
