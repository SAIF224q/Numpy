### Filtering
Want filtered data frame?

```python
powerfull_pokemon = df[df["hp"] > 150]

print(powerfull_pokemon)
```

Aggregate function: Reduces a set of values to summary value.
Used to summarise data
often used with that groupby() function.

Mean
To find the mean of columns data use .mean() function

```python
# print(df.mean()) # but our dataset contains some non numeric datas for which no mean can be calculated so set numeric_only=True

print(df.mean(numeric_only=True))
```

Sum
To find sum of column data use .sum()

```python
print(df.sum(numeric_only=True))
```

Minimum and Maximum

```python
print(f"minimum value are: \n {df.min(numeric_only=True)} \n")

print(f"maximum value are : \n {df.max(numeric_only=True)}")
```

Count

Counting number of value in each column.

```python
print(df.count())
```

How to calculate this summary for single column.

```python
print(df["speed"].mean()) # here you don't have to set numeric_only as speed is numeric data.
```

group by
this is a function that group the data with simmilar features like similar type1
but used with aggregate function.

```python
group = df.groupby("type1")
print(group["attack"].max())
```
