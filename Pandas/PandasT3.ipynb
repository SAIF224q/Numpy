Data cleaning using pandas.

The process of removing/fixing the incomplete, incorrect or irrelevent data.

~75% of work done with pandas is cleaning the data.

for this we will use original pokemon data `pokemon.csv`

```python
import pandas as pd

df = pd.read_csv("pokemon.csv")
```

```python
print(df.head())
```

1. Dropping the irrelevent columns

Let's drom some colmuns for the data we have
we have total 41 columns let's remove some and have only 5 of them.

```python
# .drop(columns = [list]) removes columns isnside the list and give new dataset.
df = df.drop(columns = ["abilities", "base_egg_steps", "base_happiness" ,"base_total"])  

print(df.head())
```

Let's drop the row which have missing values.
Column type2 have some missing values we can drop them

```python
# .dropna(subset = [list]) removes rows with missing values
df = df.dropna(subset = ["type2"])
```

```python
print(df.to_string())
```

But what is we want to replace the NA values with some other values.

For that we will use `.fillna()` and use the original data with NA type2 values.

```python
df2 = pd.read_csv("pokemon.csv")
df2 = df2.fillna({"type2" : "No Need"})

print(df2["type2"].to_string())
```

Fix incosistent values.

```python
df["type1"] = df["type1"].replace({"grass":"GRASS"})

print(df["type1"].head())
```

standardized text.

```python
df["name"] = df["name"].str.upper()

print(df["name"].to_string())
```

Fix and change the data types.

Fix - 
The is_legendary column have a 0/1 values let's change them to boolean values.

```python
df["is_legendary"] = df["is_legendary"].astype(bool)

print(df["is_legendary"].to_string())
```

Dropping the duplicate values.

```python
df = df.drop_duplicates() # we don't have any duplicates but use this function if have

# our final data set

print(df.to_string())
```
