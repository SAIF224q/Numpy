```python
# First import pandas.
import pandas as pd
```

You typically works with two type of data in pandas.
Series and Dataframe.
Series is one dimentional and DataFrame if multiDim

```python
# Data
data = [101,102,103,104]
```

Contructing series.
Series is a labled 1-dim array.

```python
series = pd.Series(data)
print(series)
```

You can see 1-dim data.
In meta data you can see we have 64 bit integer values.

But what is we have just 1 data different type like below

```python
fl = pd.Series([101.1,102])
print(fl)
```

all the other data also treated as a floating values eventhough it's int.

On the left side you are seeing the lable, by default it's index but you can change it by using an extra argument where you can pass another list, dic, or array like object.

```python
seriesLabled = pd.Series(data, index = ['a','b','c','d'])
print(seriesLabled)
```

Accessing the data points.

You can access and also manupulate the data points by using loc and iloc.
loc is for lable that you set but also if defalut lable (index) while iloc is only for by index accessing when you set the another lable but also use if used default lable.

```python
print(series.loc[0])

print(seriesLabled.loc["a"])

print(series.iloc[0])

# change/manupulate the value

seriesLabled.loc['e'] = 105

print(seriesLabled)
```

Accessing the value by condition.

```python
print(series[series < 103])
```

Using much complex data like dictionary

```python
calories = {
    "Monday": 2100,
    "Tuesday": 1950,
    "Wednesday": 2300,
    "Thursday": 1800,
    "Friday": 2200,
    "Saturday": 2500,
    "Sunday": 2400
}
```

```python
seriesCal = pd.Series(calories)

print(seriesCal)
```

```python
print(seriesCal[seriesCal < 2000])
```

```python
# I cheated on Sunday by 200 cal, let's add.

seriesCal.loc["Sunday"] += 200
print(seriesCal.loc["Sunday"])
```

DataFrame - this is our two dimentional data structure that you see in excel sheets.

```python
student_scores = {
    "Alice": [85, 90, 78, 92],
    "Bob": [70, 88, 95, 60],
    "Charlie": [100, 95, 98, 99],
    "Diana": [60, 72, 68, 75]
}
```

```python
df = pd.DataFrame(student_scores)

print(df)
```

Here each key treated as a column and list as the values in that column.

Like series you can also set a custome indexes.

```python
dfIndexed = pd.DataFrame(student_scores, index = ["English", "Hindi", "Maths", "Science"])

print(dfIndexed)
```

```python
# accessing the data

print(dfIndexed.loc["Maths"])
```

```python
# accessing a column

print(dfIndexed["Alice"])
```

```python
# adding new column

dfIndexed["Mike"] = [100,100,100,100]

print(dfIndexed)
```

adding new row.

To add new row you have to create an signle new DataFrame with all the new values for column than concatinate tham.

```python
new_rows = pd.DataFrame([{"Alice":100, "Bob":100, "Charlie":100, "Diana":100, "Mike":100},
                         {"Alice":00, "Bob":00, "Charlie":00, "Diana":00, "Mike":100}]
                         , index=["IT","IP"])

new_dfIndexed = pd.concat([dfIndexed, new_rows])

print(new_dfIndexed)
```
